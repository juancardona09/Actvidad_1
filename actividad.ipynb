# Actividad 1: carga y resumen de datos con Hugging Face

En este notebook se cargan los datos de `scikit-learn/iris` con la clase `HFDataLoader`, se aplica una transformacion minima (filtrar filas) y se genera el resumen con la clase `DatasetSummarizer`.

```python
from hf_data_loader import HFDataLoader
from dataset_summarizer import DatasetSummarizer
```

```python
# 1. Cargamos el dataset de Hugging Face como DataFrame
loader = HFDataLoader("scikit-learn/iris", split="train")
df = loader.load()

df.head()
```

```python
# 2. Transformacion minima: filtramos las filas con petalos largos
df_filtrado = df[df["PetalLengthCm"] > 1.5]

df_filtrado.head()
```

**Transformacion aplicada:** se filtraron las filas donde `PetalLengthCm` es mayor a 1.5.

El resultado se guarda en `df_filtrado`, una variable nueva. Por eso el resumen tiene menos filas que el dataset original.

```python
# 3. Pasamos el DataFrame transformado a la clase de resumen
resumidor = DatasetSummarizer("resumen_dataset.txt")
texto = resumidor.summarize(df_filtrado)

print(texto)
```
